# Maricopa County ZIP Regression

```python
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
```

```python
data = pd.read_csv('maricopa_zip_merged.csv')
data = data.dropna()
data = data[data['median_home_value'] < 2000000]
len(data)
```

```python
data.describe()
```

```python
data[['median_household_income','pct_bachelors','median_year_built']].corr().round(2)
```

## Model 1

```python
X = sm.add_constant(data[['median_household_income','pct_bachelors','median_year_built']])
m1 = sm.OLS(data['median_home_value'], X).fit()
print(m1.summary())
```

## Model 2

```python
m2 = sm.OLS(data['median_gross_rent'], X).fit()
print(m2.summary())
```
